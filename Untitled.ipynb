```python
import torch
from torchvision import datasets, transforms
```

```python
train_dataset = datasets.MNIST(root='.',
   download=True,
   train=True,
   transform=transforms.ToTensor()
)

                       
test_dataset = datasets.MNIST(
    root='.',
    train=False,                       
    download=True,
    transform=transforms.ToTensor()
)
```

```python
xnum = []
ynum = []
for example in train_dataset:
    xnum.append(example[0])
    ynum.append(example[1])
xnum = torch.stack(xnum)
xnum = torch.reshape(xnum, (60000,784))
ynum = torch.tensor(ynum)
xnum.shape
```

```python
hidden_layer = 100
b1 = torch.randn(hidden_layer, requires_grad=True)
W1 = torch.randn((xnum.shape[1], hidden_layer), requires_grad=True)
h = torch.tanh(xnum @ W1 + b1)
h.shape
```

```python
#100, 10
b2 = torch.randn(10, requires_grad=True)
W2 = torch.randn((hidden_layer, 10), requires_grad=True)
logits = h @ W2 + b2
counts = logits.exp()
probs = counts/counts.sum(1, keepdim=True)
print(probs)
```

```python
parameters = [W1, b1, W2, b2]
lr = 0.1
for i in range(100):
    #forward pass
    h = torch.tanh(xnum @ W1 + b1)
    logits = h @ W2 + b2 
    counts = logits.exp()
    probs = counts/counts.sum(1, keepdim=True)
    loss = -probs[torch.arange(len(ynum)), ynum].log().mean()
    #backward pass
    for p in parameters:
        p.grad = None
    loss.backward()
    for p in parameters:
        p.data += -lr * p.grad
    print(loss.item()) 
```
