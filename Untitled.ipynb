```python
import torch
import matplotlib.pyplot as plt
from torchvision import datasets, transforms
```

```python
train_dataset = datasets.MNIST(root='.',
   download=True,
   train=True,
   transform=transforms.ToTensor()
)

                       
test_dataset = datasets.MNIST(
    root='.',
    train=False,                       
    download=True,
    transform=transforms.ToTensor()
)
```

```python
xnum = []
ynum = []
for example in train_dataset:
    xnum.append(example[0])
    ynum.append(example[1])
xnum = torch.stack(xnum)
xnum = torch.reshape(xnum, (60000,784))
ynum = torch.tensor(ynum)
xnum.shape
```

```python
hidden_layer = 100
b1 = torch.randn(hidden_layer, requires_grad=True)
W1 = torch.randn((xnum.shape[1], hidden_layer), requires_grad=True)
h = torch.tanh(xnum @ W1 + b1)
h.shape
```

```python
#100, 10
b2 = torch.randn(10, requires_grad=True)
W2 = torch.randn((hidden_layer, 10), requires_grad=True)
logits = h @ W2 + b2
counts = logits.exp()
probs = counts/counts.sum(1, keepdim=True)
print(probs)
```

```python
batch_size = 128
parameters = [W1, b1, W2, b2]
lr = 0.05
for i in range(1000):
    bix = torch.randint(0, xnum.shape[0], (batch_size,))
    #forward pass
    h = torch.tanh(xnum[bix] @ W1 + b1)
    logits = h @ W2 + b2 
    counts = logits.exp()
    probs = counts/counts.sum(1, keepdim=True)
    loss = -probs[torch.arange(batch_size), ynum[bix]].log().mean()
    #backward pass
    for p in parameters:
        p.grad = None
    loss.backward()
    for p in parameters:
        p.data += -lr * p.grad
    print(loss.item()) 
```

```python
xtest = []
ytest = []
for example in test_dataset:
    xtest.append(example[0])
    ytest.append(example[1])

xtest = torch.stack(xtest)
xtest = torch.reshape(xtest, (10000, 784))
ytest = torch.tensor(ytest)

with torch.no_grad():
    h_test = torch.tanh(xtest @ W1 + b1)
    logits_test = h_test @ W2 + b2

preds = torch.argmax(logits_test, dim=1)
accuracy = (preds == ytest).float().mean()
print(accuracy.item())
```

```python
idx = 35
image_to_show = xtest[idx].reshape(28, 28)

plt.imshow(image_to_show, cmap='gray')
plt.title(f"y: {ytest[idx].item()}, pred {preds[idx].item()}")
plt.axis('off')
plt.show()
```
